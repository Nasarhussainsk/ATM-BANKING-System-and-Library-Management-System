```python
from abc import ABC, abstractmethod


class Secureuser(ABC):

    def __init__(self, name, atm_pin):
        self.name = name
        self.__atm_pin = atm_pin
        self._balance = 0

    def get_pin(self):
        for i in range(3):
            attempt = int(input("Enter your ATM PIN: "))

            if attempt == self.__atm_pin:
                return "PIN IS CORRECT"
            else:
                chances = 2 - i

                if chances > 0:
                    print(f"PIN IS INCORRECT. YOU HAVE {chances} CHANCES LEFT.")
                else:
                    print("Your account is blocked")
                    return "Access denied"

    @abstractmethod
    def deposit_cash(self, amount):
        pass

    @abstractmethod
    def withdraw_cash(self, amount):
        pass

    def account_balance(self):
        return f"Your account balance is ₹{self._balance}"
        
class SavingsAccount(Secureuser):

    def deposit_cash(self, amount):
        self._balance += amount
        return f"₹{amount} deposited successfully. Balance: ₹{self._balance}"

    def withdraw_cash(self, amount):
        if amount <= self._balance:
            self._balance -= amount
            return f"₹{amount} withdrawn successfully. Balance: ₹{self._balance}"
        return "Insufficient Balance"
        
class CurrentAccount(Secureuser):

    def deposit_cash(self, amount):
        self._balance += amount
        return f"₹{amount} deposited into current account. Balance: ₹{self._balance}"

    def withdraw_cash(self, amount):
        if amount <= self._balance:
            self._balance -= amount
            return f"₹{amount} withdrawn from current account. Balance: ₹{self._balance}"
        return "Insufficient Balance"
```

```python
savings = SavingsAccount("john", 1234)
current = CurrentAccount("rio", 5678)

print(savings.deposit_cash(9000))
print(savings.withdraw_cash(1000))

print(current.deposit_cash(10000))
print(current.withdraw_cash(9000))
```

```python
from abc import ABC, abstractmethod
class LibraryUser(ABC):
    def __init__(self,name,user_id):
        self.name=name
        self.__user_id=user_id
        self._borrowed_books=[]
    def get_id(self):
        for i in range(3):
            attempt=int(input("Enter Your I'd:"))
            if attempt ==self.__user_id:
                return "Your I'd is successfully verified"
            else:
                chances = 2 - i

                if chances > 0:
                    print(f"PIN IS INCORRECT. YOU HAVE {chances} CHANCES LEFT.")
                else:
                    print("Your account is blocked")
                    return "Access denied"
    @abstractmethod
    def borrow_book(self,book):
        pass
    @abstractmethod
    def return_book(self,book):
        pass
        
    def view_borrowed_books(self):
        return f"This is the list of borrowed books,{self._borrowed_books}"

class StudentUser(LibraryUser):
    def borrow_book(self,book):
        if len(self._borrowed_books)<3:
            self._borrowed_books.append(book)
            return f"you have borrowed '{book}'.Total borrowed:{len(self._borrowed_books)}"
        else:
            return "Limit Reached.Students can borrow only 3 books"
    def return_book(self,book):
        if book in self._borrowed_books:
            self._borrowed_books.remove(book)
            return f"You have returned a '{book}'. Remaining:{len(self._borrowed_books)}"
        else:
            return f"You don't have'{book}'borrowed."
    
class TeacherUser(LibraryUser):
    def borrow_book(self,book):
        if len(self._borrowed_books) < 5:
            self._borrowed_books.append(book)
            return f"you have borrowed {book}. Total borrowed:{len(self._borrowed_books)}"
        else:
            return "Limit Reached.Teachers can borrow only 5 books"
    def return_book(self,book):
        if book in self._borrowed_books:
            self._borrowed_books.remove(book)
            return f"You have returned a '{book}'. Remaining:{len(self._borrowed_books)}"
        else:
            return f"You don't have'{book}'borrowed."
```

```python
student1 = StudentUser("Ali", 102)
teacher1 = TeacherUser("Mrs. Smith", 201)

print(student1.borrow_book("Math Book"))
print(student1.borrow_book("Science Book"))
print(student1.borrow_book("History Book"))
print(student1.borrow_book("English Book"))  # Limit reached

print(teacher1.borrow_book("Physics Book"))
print(teacher1.borrow_book("Chemistry Book"))
print(teacher1.borrow_book("Biology Book"))
print(teacher1.borrow_book("Math Book"))
print(teacher1.borrow_book("English Book"))
print(teacher1.borrow_book("Extra Book"))    # Limit reached
```
